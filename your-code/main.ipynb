## Bernoulli's Distribution

In a basket containing 100 fruits, there are 60 apples and 40 oranges. 

A fruit is picked randomly from the basket. 

- p **What is the probability that the fruit is apple?**
- q **What is the probability that it is an orange?**

````python
# p = probability that the fruit is an apple 
p = 60/100

# q = probability that the fruit is an orange
q = 40/100
````

Now we take a random sample of 20 fruits from the basket. After each fruit is taken, a new fruit of the same type is replaced in the basket. Therefore, every time we are taking 1 fruit from 100 fruits. 

1. Q1: **What is the probability that the first 5 fruits are all apples?**

1. Q2: **What is the probability that the first 5 fruits are all apples and the next 15 fruits are all oranges?**

You can include the `p` and `q` probabilities you previous calculated in your solution.

````python
"""

What is the probability that the first 5 fruits are all apples?

The individual Bernoulli trial is: taking a fruit and that fruit being apple. As a single event, the probability of that event is equal to p. 

But because we want all first five to be apples, then this is the joint probability: 
"""
Q1 = p**5

print(f"The probability of the first five fruits being apple is {Q1:.2f} out of 1")
````

````python
'''
This is also a problem where the probabilities should be joined, but in this case the probabilities of Q1 and now Q2 which would be similar:

'''

Q2 = Q1*(q**15)

print(f"The probability of the first five fruits being apples and the next 15 being orange is {Q2} out of 1")
````

## Binomial Distribution

Similar to the above problem, this time again we randomly sample 20 fruits.

Q3: **What is the probability that this time the sample contains 5 apples and 15 oranges?**

Please note that this time the order of the fruits being picked does not matter.

````python
from scipy.stats import binom
````

````python
#This requires the binomial function.
#Sample is 
n = 20
#The event implies choosing 5 apples
x = 5
#The probability of 5 fruits being apples will be A
A = binom.pmf(x,n,p)

#The second event implies choosing 15 oranges. 
y = 15
#The probability of 15 fruits being oranges is B: 
B = binom.pmf(y,n,q)
````

````python
#Because the events are independent, they can be calculated into a joint probability: 
Q3 = A*B
print(f"The probability of the five fruits being apples and 15 being oranges is {Q3} out of 1")
````

In the lesson we also discussed CDF (cumulative distribution function). In the context of the above problem, **what is the probability that less than 5 fruits picked are apples?**

Write the formula and calculate the probability below.

````python
'''
The sci.py function of binom.cdf(m,N,p) calculates the lower tail probability
by default (P(X<x) so this should answer the question
'''

Q4 = binom.cdf(x,n,p)
Q4
print(f"The probability of that less than five fruits are apples is {Q4}, compared to when exactly 5 where applest that was {A}, both out of 1")
````

**Plot the Probability Mass Function (PMF) for the above binomial distribution.**

*Hint: The range of x in a binomial distribution is from `0` to `n`.*

````python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
````

````python
#Plotting binomial distributions A and B. 
#The outcome of the binomial is a single string, so we would have to calculate the distributions for each scenario,from 0 to 20. 
rangeX= np.arange(0,21) 
rangeY= np.arange(0,21)

#Now I take k as a range, rather than a single value, so it outputs the array of pmfs.
pmf_A = binom.pmf(rangeX,n,p)
pmf_B = binom.pmf(rangeY,n,q)


sns.barplot(x= rangeX, y = pmf_A, alpha=0.5, color = "red", legend= True)
sns.barplot(x= rangeY, y = pmf_B, alpha=0.5, color = "orange", legend = True)
plt.title("Binomial pmf distribution")
plt.xlabel("Number of fruits")
plt.ylabel("Probability out of 1")
plt.legend(["Apples", "Oranges"])
plt.show()
````

## Poisson Distribution

In this challenge you are required to use the Math library in python to solve the problems.

In the UEFA champions league tournament, the average number of goals scored in a match is 2.3. 

**What is the probability that the number of goals scored in a match is 5?**

*Hint: Use the exp() in the math library*

And to calculate the factorial, you can use the following code:

```python
import math
math.factorial(n) # n is the number you want to find the factorial of
```

````python
import math
````

````python
#The formula of the Poisson distributtion is: (Lamda^k)*(e^-lambda)/k!
L = 2.3 #Lamda: goals per match
k = 5
PD = ((L**k)*math.exp(-L))/math.factorial(k)
print(f"The probability of scoring 5 goals per match is {PD:.3f} out of 1.")
````

**Draw a poisson probability distribution plot for the number of goals to be between 0 to 10.**

````python
from scipy.stats import poisson
````

````python
rangeG=np.arange(0,11)
pmf_PD = poisson.pmf(rangeG,L)
sns.barplot(x= rangeG, y = pmf_PD, alpha=0.5, color = "red")
plt.title("Poisson pmf distribution")
plt.xlabel("Goals per match")
plt.ylabel("Probability out of 1")
plt.show()
````
